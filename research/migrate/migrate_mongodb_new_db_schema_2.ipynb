# Law social relation

```python
import pymongo
from pymongo import UpdateOne
from concurrent.futures import ThreadPoolExecutor


def mapping_document(old_doc):
    """
    Chuyển đổi document từ schema cũ sang schema law_social_relation mới.
    """
    raw_status = str(old_doc.get("status", "")).strip().upper()
    status = "ACTIVE" if raw_status == "ACTIVE" else "INACTIVE"

    new_doc = {
        "social_relation_id": old_doc.get("social_relation_id", ""),
        "social_relation_name": old_doc.get("social_relation_name", ""),
        "social_relation_name_norm": old_doc.get("social_relation_name_norm", ""),
        "status": status,
        "created_at": old_doc.get("created_date", ""),
        "created_by": old_doc.get("created_by", ""),
        "last_modified_at": old_doc.get("last_modified", ""),
        "last_modified_by": old_doc.get("last_modified_by", "")
    }
    return new_doc

def process_batch(batch_docs):
    """
    Hàm xử lý một nhóm document và ghi vào target bằng Bulk Write.
    """
    if not batch_docs:
        return
    
    target_collection = client[TARGET_CORE_DB]["law_social_relation"]
    operations = []
    
    for doc in batch_docs:
        new_doc = mapping_document(doc)
        # Tạo lệnh update/upsert
        op = UpdateOne(
            {"social_relation_id": new_doc["social_relation_id"]},
            {"$set": new_doc},
            upsert=True
        )
        operations.append(op)
    
    if operations:
        target_collection.bulk_write(operations, ordered=False)

def main_migration():
    # 1. Load dữ liệu
    source_collection = client[SOURCE_CORE_DB]["law_social_relation"]
    source_collection_2 = client[SOURCE_CORE_DB]["law_social_relation_draft"]
    
    print("Reading documents...")
    documents = list(source_collection.find()) + list(source_collection_2.find())
    total_docs = len(documents)
    
    # 2. Chia nhỏ dữ liệu thành các batch (ví dụ: mỗi batch 500 records)
    batch_size = 1000
    batches = [documents[i:i + batch_size] for i in range(0, total_docs, batch_size)]
    
    print(f"Starting multithreaded migration for {total_docs} documents...")

    # 3. Sử dụng ThreadPoolExecutor để chạy đa luồng
    # max_workers có thể điều chỉnh tùy theo cấu hình CPU/Network (thường từ 4-10)
    with ThreadPoolExecutor(max_workers=8) as executor:
        executor.map(process_batch, batches)

    print(f"✅ Migrated {total_docs} documents to {TARGET_CORE_DB}.law_social_relation")

if __name__ == "__main__":
    main_migration()
```

# Law social relation mapping + draft

```python
import pymongo
from pymongo import UpdateOne
from concurrent.futures import ThreadPoolExecutor

# Giả định client, SOURCE_CORE_DB, TARGET_CORE_DB đã được định nghĩa trước đó

def mapping_document(old_doc):
    """
    Chuyển đổi document từ collection gốc sang collection law_social_relation_mapping mới.
    """
    if not old_doc:
        return {}

    # 1. Chuẩn hóa Relation Type: "Primary" -> "PRIMARY"
    raw_relation_type = old_doc.get("relation_type", "")
    relation_type = str(raw_relation_type).strip().upper() if raw_relation_type else ""

    # 2. Xây dựng document mới theo schema chuẩn hóa
    new_doc = {
        "doc_id": str(old_doc.get("doc_id", "")),
        "article_id": str(old_doc.get("article_id", "")),
        "social_relation_id": str(old_doc.get("social_relation_id", "")),
        "relation_type": relation_type,
        "created_at": str(old_doc.get("created_date", "")),
        "created_by": str(old_doc.get("created_by", "")),
        "last_modified_at": str(old_doc.get("last_modified", "")),
        "last_modified_by": str(old_doc.get("last_modified_by", ""))
    }
    return new_doc

def process_batch(batch_docs):
    """
    Xử lý một nhóm document và thực hiện bulk_write vào target collection.
    """
    if not batch_docs:
        return
    
    target_collection = client[TARGET_CORE_DB]["law_social_relation_mapping"]
    operations = []
    
    for doc in batch_docs:
        new_doc = mapping_document(doc)
        
        # Lưu ý quan trọng về Filter: 
        # Vì đây là bảng mapping, một social_relation_id có thể đi với nhiều doc_id/article_id.
        # Nên filter cần kết hợp cả 3 trường để tránh ghi đè nhầm dữ liệu.
        filter_criteria = {
            "doc_id": new_doc["doc_id"],
            "article_id": new_doc["article_id"],
            "social_relation_id": new_doc["social_relation_id"]
        }
        
        operations.append(
            UpdateOne(filter_criteria, {"$set": new_doc}, upsert=True)
        )
    
    if operations:
        try:
            target_collection.bulk_write(operations, ordered=False)
        except Exception as e:
            print(f"❌ Lỗi khi thực hiện bulk write: {e}")

def main_migration():
    # 1. Load dữ liệu từ 2 nguồn
    print("🚀 Đang tải dữ liệu từ source...")
    source_collection = client[SOURCE_CORE_DB]["law_social_relation_mapping"]
    source_collection_2 = client[SOURCE_CORE_DB]["law_social_relation_mapping_draft"]
    
    documents = list(source_collection.find()) + list(source_collection_2.find())
    total_docs = len(documents)
    print(f"📦 Tổng số bản ghi cần migrate: {total_docs}")

    # 2. Cấu hình Batch và Thread
    batch_size = 1000  # Mỗi thread xử lý 1000 bản ghi một lúc
    max_workers = 10   # Số lượng thread chạy song song
    
    # Chia nhỏ danh sách documents thành các batch
    batches = [documents[i:i + batch_size] for i in range(0, total_docs, batch_size)]

    # 3. Thực thi đa luồng
    print(f"⚙️ Đang migrate với {max_workers} threads...")
    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        executor.map(process_batch, batches)

    print(f"✅ Migrated {total_docs} documents to {TARGET_CORE_DB}.law_social_relation_mapping thành công!")

if __name__ == "__main__":
    main_migration()
```

```python
# Load documents from source collection
source_collection = client[MID_CORE_DB]["law_tree"]
documents = list(source_collection.find())
# Define function to migrate documents

from datetime import datetime

def mapping_document(doc: dict) -> dict:
    """
    Map document từ collection cũ sang schema law_tree
    """

    def normalize(value):
        """Chuẩn hóa dữ liệu thiếu -> ''"""
        if value is None:
            return ""
        return value

    def normalize_status(status):
        """Chuẩn hóa status -> ACTIVE | INACTIVE"""
        if not status:
            return ""
        status = str(status).upper()
        if status not in ["ACTIVE", "INACTIVE"]:
            return ""
        return status

    def normalize_datetime(value):
        """
        Chuẩn hóa datetime về string '%Y-%m-%d %H:%M:%S'
        Hỗ trợ Mongo $date hoặc string
        """
        if value is None:
            return ""

        if isinstance(value, dict) and "$date" in value:
            try:
                dt = datetime.fromisoformat(value["$date"].replace("Z", ""))
                return dt.strftime("%Y-%m-%d %H:%M:%S")
            except Exception:
                return value["$date"]

        if isinstance(value, datetime):
            return value.strftime("%Y-%m-%d %H:%M:%S")

        return str(value)

    def normalize_int(value):
        """Chuẩn hóa số nguyên"""
        try:
            return int(value)
        except Exception:
            return 0

    new_doc = {
        "tree_id": normalize(doc.get("tree_id")),
        "tree_name": normalize(doc.get("tree_name")),
        "count": normalize_int(doc.get("count")),
        "category": normalize(doc.get("category")),
        "state": normalize(doc.get("state")),
        "status": normalize_status(doc.get("status")),
        "created_at": normalize_datetime(doc.get("created_at")),
        "created_by": normalize(doc.get("created_by")),
        "last_modified_at": normalize_datetime(doc.get("last_modified")),
        "last_modified_by": normalize(doc.get("modified_by")),
    }

    return new_doc
# Migrate documents to target collection
target_collection = client[TARGET_CORE_DB]["law_tree"]
for doc in documents:
    new_doc = mapping_document(doc)
    target_collection.update_one({"tree_id": new_doc["tree_id"]}, {"$set": new_doc}, upsert=True)

print(f"✅ Migrated {len(documents)} documents to {TARGET_CORE_DB}.law_tree")
```
